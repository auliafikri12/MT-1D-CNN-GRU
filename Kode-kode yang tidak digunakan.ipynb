```python
#Fungsi Pembuatan Dataset
def generate_ketebalan(n_layers):
    k = n_layers - 1  # Jumlah ketebalan
    total = 990  # Total ketebalan (10k ≤ total ≤ 990)
    ketebalan = []
    remaining = total
    for i in range(k - 1):
        max_val = remaining - 10 * (k - i - 1)  # Maksimal ketebalan per layer
        layer = np.random.choice(np.arange(10, max_val + 1, 10))  # Pilih ketebalan
        ketebalan.append(layer)
        remaining -= layer
    ketebalan.append(remaining)  # Sisa untuk layer terakhir
    return np.array(ketebalan)
```
